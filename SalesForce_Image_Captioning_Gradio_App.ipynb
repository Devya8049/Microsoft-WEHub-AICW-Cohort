```python

from transformers import pipeline
import gradio as gr

# Load the BLIP image captioning model
print("Loading BLIP model... Please wait.")

pipe = pipeline(
    "image-to-text",
    model="Salesforce/blip-image-captioning-large"
)

print("Model loaded successfully!")


# Function to generate image caption
def generate_caption(image):
    if image is None:
        return "Please upload an image."

    result = pipe(image, max_new_tokens=100)

    caption = result[0]["generated_text"]

    return caption


# Create Gradio interface
demo = gr.Interface(
    fn=generate_caption,
    inputs=gr.Image(type="pil", label="Upload an Image"),
    outputs=gr.Textbox(label="Generated Image Caption"),
    title="AI Image Caption Generator",
    description=(
        "Upload an image and the BLIP model "
        "will generate a natural-language description."
    ),
    examples=None
)

# Launch the application
if __name__ == "__main__":
    demo.launch()
```
