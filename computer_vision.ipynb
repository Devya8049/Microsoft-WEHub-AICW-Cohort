```python
#SPLITTING A COLOR IMAGE INTO 3 CHANNELS
bird=cv2.imread('D:/8049/image2.jpg',cv2.IMREAD_COLOR) #BGR
b,g,r=cv2.split(bird) #BGR
plt.figure(figsize=[20,5])
plt.subplot(1,5,1);plt.imshow(b,cmap='gray');plt.title('Blue Channel')
plt.subplot(1,5,2);plt.imshow(g,cmap='gray');plt.title('Green Channel')
plt.subplot(1,5,3);plt.imshow(r,cmap='gray');plt.title('Red Channel')
img_merge=cv2.merge((b,g,r))
plt.subplot(1,5,4);plt.imshow(img_merge[:,:,::-1]);plt.title('Merged Channel')
```

```python
# reading original image
flower_BGR=cv2.imread("D:/8049/image5.png") #BGR
sun_RGB=cv2.cvtColor(flower_BGR,cv2.COLOR_BGR2RGB)
plt.imshow(sun_RGB)
```

```python
# CONVERT INTO GRAY SCALE IMAGE
sun_gray=cv2.cvtColor(sun_RGB,cv2.COLOR_RGB2GRAY)
plt.imshow(sun_gray,cmap='gray')
```

```python
print(sun_RGB.shape)
print(sun_gray.shape)
```

```python
print(sun_gray)
```

```python
sun_blur=cv2.GaussianBlur(sun_gray,(3,3),0)
plt.imshow(sun_blur,cmap='gray')
print(sun_blur)
```

```python
checker=cv2.imread('D:/8049/Image1.jpeg',0)
plt.imshow(checker,cmap='gray')
print(checker)
print(checker.shape)
```

```python
checker_blur=cv2.GaussianBlur(checker,(3,3),0)
plt.imshow(checker_blur,cmap='gray')
print(checker_blur)
```

```python
# Edge Direction - Canny Edge Detection
sun_edge=cv2.Canny(sun_gray,75,75)
plt.imshow(sun_edge)
```

```python
# social edge dectection
sunx=cv2.Sobel(sun_blur,ddepth=cv2.CV_64F,
               dx=1,dy=0,ksize=5)
suny=cv2.Sobel(sun_blur,ddepth=cv2.CV_64F,
               dx=0,dy=1,ksize=5)
sunxy=cv2.Sobel(sun_blur,ddepth=cv2.CV_64F,
               dx=1,dy=1,ksize=5)
plt.figure(figsize=[18,5])
plt.subplot(1,4,1);plt.imshow(sunx,cmap='gray');plt.title('Sunx')
plt.subplot(1,4,2);plt.imshow(suny,cmap='gray');plt.title('Suny')
plt.subplot(1,4,3);plt.imshow(sunxy,cmap='gray');plt.title('Sunxy')
```

```python
# threshold image - black and white image
ret_val,threshold_img=cv2.threshold(sun_gray,110,255,cv2.THRESH_BINARY)
plt.imshow(threshold_img,cmap='gray')
print(threshold_img)
```
