```python
import numpy as np
import matplotlib.pyplot as plt

student = np.array(["Alice", "Bob", "Charlie", "David", "Eve"])
marks = np.array([85, 72, 90, 65, 78])

plt.bar(student, marks)
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Bar Graph")

plt.show()
```

```python
marks = np.array([45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 82, 85, 88, 90, 92, 95])

plt.hist(marks, bins=5, edgecolor="black")

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Students' Marks")

plt.show()
```

```python
marks = np.array([45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95])

plt.boxplot(marks)

plt.ylabel("Marks")
plt.title("Boxplot of Students' Marks")

plt.show()
```

```python
maths = np.array([65, 70, 75, 80, 85, 90, 95])
science = np.array([60, 68, 72, 78, 82, 88, 92])
english = np.array([70, 74, 78, 80, 84, 89, 94])

data = [maths, science, english]

plt.boxplot(data, tick_labels=["Maths", "Science", "English"])

plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.title("Comparison of Marks")

plt.show()
```

```python
fruits = np.array(["Apple", "Banana", "Mango", "Orange"])
y = np.array([30, 25, 20, 25])

plt.pie(y, labels=fruits)
plt.title("Fruit Distribution")
plt.legend()
plt.show()
```

```python
sns.set_theme()
tips=sns.load_dataset("tips")
sns.relplot(
    data=tips,
    x="total_bill", y="tip", col="time",
    hue="smoker", style="smoker", size="size"
)
```

```python
rs= np.random.RandomState(10)
d=rs.normal(size=100)
colors=["r","g","b"]
#hist
sns.distplot(
    d,kde=True,hist=True,bins=10,rug=True,
    hist_kws={"alpha":0.3,"color":colors[0]}, #r
    kde_kws={"color":colors[1],"lw":2}, #g
    rug_kws={"color":colors[2]} #b
)
```

```python
rs= np.random.RandomState(10)
d=rs.normal(size=100)
colors=["r","g","b"]
#hist
sns.distplot(
    d,kde=True,hist=True,bins=20,rug=True,
    hist_kws={"alpha":0.3,"color":colors[0]}, #r

)
```
