```python
# Step 1: Tokenization
from nltk.tokenize import sent_tokenize, word_tokenize
text="""There are multiple ways! we can perform 'tokenization' on a given text data. We can choose any method based on language, library, and purpose of modelling."""
sent=sent_tokenize(text)
print(sent)
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
print(stopwords.words('english')) # stopwords in eng language
print(len(stopwords.words('english')))
```

```python
print(stopwords.words('french')) # stopwords in eng language
print(len(stopwords.words('french')))
```

```python
words=['There', 'are', 'multiple', 'ways', '!', 'We', 'can', 'perform', "'tokenization", "'", 'on', 'given', 'text', 'data', '.', 'We', 'can', 'choose', 'any', 'method', 'based', 'on', 'language', ',', 'library', ',', 'and', 'purpose', 'and', 'modelling', '.']
```

```python
stop_words=set(stopwords.words('english'))
filtered_words=[]
for w in words:
    if w not in stop_words:
        filtered_words.append(w)
print(filtered_words)
```

```python
print("\n Original Sentence:")
print(" ".join(words))
print("\n Filtered Sentence (after stop-word removal):")
print(" ".join(filtered_words))
```

```python
from nltk.stem import PorterStemmer
ps= PorterStemmer()
```

```python
word_input=["leaving","building","pythoning","best","better","pythoner","pythonly","Stemming"]
for w in word_input:
    print(f"{w}-->{ps.stem(w)}")
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet') # uses this dictionary
lem= WordNetLemmatizer()
```

```python
for w in word_input:
    print(f"{w}---->{lem.lemmatize(w,pos='v')}")
```

```python
for w in word_input:
    print(f"{w}---->{lem.lemmatize(w,pos='a')}")
```

```python
print(lem.lemmatize('stemming',pos='v'))
print(lem.lemmatize('flying',pos='v'))
print(lem.lemmatize('better',pos='a'))
print(lem.lemmatize('building',pos='v'))
print(lem.lemmatize('building',pos='n'))
```
