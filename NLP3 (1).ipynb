```python
# Use 'text-generation' pipeline as 'summarization' has been deprecated in Transformers v5
generator = pipeline("text-generation", model="gpt2")

# Combine the prompt and text for summarization
prompt = f"Summarize the following text:\n{text}\n\nSummary:"
summary = generator(prompt, max_new_tokens=50, num_return_sequences=1)
print(summary[0]['generated_text'])
```
